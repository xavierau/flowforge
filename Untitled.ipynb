```python
import pdfplumber

def extract_features(pdf_path, page_index=1):
    tokens = []
    boxes = []

    with pdfplumber.open(pdf_path) as pdf:
        page = pdf.pages[page_index-1]
        width, height = page.width, page.height

        # Extract words and their bounding boxes
        for word in page.extract_words():
            text = word['text']

            # Normalize coordinates to 0-1000 scale
            x0 = int((word['x0'] / width) * 1000)
            y0 = int((word['top'] / height) * 1000)
            x1 = int((word['x1'] / width) * 1000)
            y1 = int((word['bottom'] / height) * 1000)

            # Safety clamp (sometimes PDFs have weird hidden text off-page)
            box = [
                max(0, min(1000, x0)),
                max(0, min(1000, y0)),
                max(0, min(1000, x1)),
                max(0, min(1000, y1))
            ]

            tokens.append(text)
            boxes.append(box)

    return tokens, boxes
```
