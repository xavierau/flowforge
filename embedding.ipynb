```python
from google import genai
import os
from dotenv import load_dotenv
load_dotenv()

client = genai.Client(api_key=os.environ['GOOGLE_API_KEY'])
```

```python
contents = [
"Implement robust end-to-end integration tests and enforce strong data typing to ensure consistency across all financial transaction data creation, storage, and retrieval processes.",
    "Always ensure data read operations are consistent with data write operations, particularly when changing underlying data structures or business logic, and include comprehensive data migration for historical records.",
    "Implement rigorous data contract enforcement, such as using enums and database constraints for critical fields, and ensure comprehensive integration testing to prevent discrepancies in data definition and usage across system components and over time.",
    "To prevent data inconsistencies and reporting errors, strictly enforce consistent data definitions across all system layers and implement thorough integration tests, including data migration scenarios, when critical data flows or schemas evolve.",
    "Always validate consistency between data creation and retrieval logic, and implement necessary data migrations for existing data when changing core system behaviors.",
    "this is just a complete out of scope text",
    "Never validate consistency between data creation and fetch. "
]


result = client.models.embed_content(
        model="gemini-embedding-001",
        contents=contents)

print(result.embeddings)
```

```python
# compare similarity
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from scipy.spatial.distance import euclidean
```

```python
# Define two example vectors
vector1 = np.array(result.embeddings[0].values)
vector2 = np.array(result.embeddings[6].values)

# Calculate cosine similarity
similarity_score = cosine_similarity(vector1.reshape(1, -1), vector2.reshape(1, -1))[0][0]
print(f"Cosine Similarity: {similarity_score}")
```
